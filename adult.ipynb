```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x="income", hue="income", legend=False)

plt.title("Number of People in Each Income Category")
plt.xlabel("Income")
plt.ylabel("Number of People")
plt.tight_layout()
plt.show()
```

The chart shows the number of people in each income catogory. Most people in the dataset earn 50K or less, while a smaller number earn more than 50K. This shows that the dataset is imbalanced because the number of people in the two income groups is different.

```python
plt.figure(figsize=(8, 5))
sns.histplot(
    data=df,
    x="age",
    hue="income",
    bins=30,
    multiple="layer",
    alpha=0.5
)

plt.title("Age Distribution by Income Category")
plt.xlabel("Age")
plt.ylabel("Number of People")
plt.tight_layout()
plt.show()
```

The histogram shows the age distribution of people in different income categories. Most people earning 50K or less are younger, while the number of people earning more than 50K is higher among older age groups. This suggests that age may have a relationship with income.

```python
plt.figure(figsize=(7, 5))
sns.boxplot(data=df, x="income", y="hours-per-week", hue="income", legend=False)

plt.title("Weekly Working Hours by Income Category")
plt.xlabel("Income")
plt.ylabel("Hours Worked per Week")
plt.tight_layout()
plt.show()
```

The boxplot shows the relationship between weekly working hours and income categories. People earning more than 50K generally work slightly longer hours per week. Their working hours also show greater variation compared to people earning 50K or less. However, both income groups have some unusual values (outliers). This suggests that working hours may have a relationship with income.

```python
plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    x="education",
    hue="income",
    order=df["education"].value_counts().index
)

plt.title("Education Level by Income Category")
plt.xlabel("Education Level")
plt.ylabel("Number of People")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

The chart shows the relationship between education level and income category. Most people in the dataset have a high school education or some college education. People with Bachelor's, Master's and Doctorate degrees have a higher representation in the over $50K income category compared to some lower education groups. This suggests that higher education may be associated with higher income.

```python
# Select numerical columns
numerical_columns = df.select_dtypes(include="number").columns

# Calculate descriptive statistics
descriptive_stats = df[numerical_columns].agg(
    ['mean', 'median', 'std']
).T

# Calculate mode
descriptive_stats['mode'] = df[numerical_columns].mode().iloc[0]

# Display results
descriptive_stats
```

 Descriptive Statistics – Interpretation

The descriptive statistics show that the average age of individuals in the dataset is approximately 38 years, while the median age is 37. The average working time is around 41 hours per week, with a median of 40 hours. This suggests that many individuals work close to a standard full-time working week.

The education-num variable has a mean of 10.12 and a median of 10, showing that the education levels are relatively centred around this value. However, capital-gain and capital-loss both have a median of zero. This indicates that at least half of the observations have no recorded capital gain or capital loss. Their high standard deviations suggest that the values vary considerably.

Overall, the results show differences in the distribution of numerical variables. These findings will help identify patterns and select suitable variables for further statistical analysis.
